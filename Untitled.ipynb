```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error
import warnings
warnings.filterwarnings('ignore')

df = pd.read_csv('superstore.csv', encoding='latin-1')
print(df.shape)
print(df.head())
```

```python
# Parse dates
df['Order Date'] = pd.to_datetime(df['Order Date'])

# Aggregate sales by month
df['YearMonth'] = df['Order Date'].dt.to_period('M')
monthly_sales = df.groupby('YearMonth')['Sales'].sum().reset_index()
monthly_sales['YearMonth'] = monthly_sales['YearMonth'].dt.to_timestamp()
monthly_sales = monthly_sales.sort_values('YearMonth').reset_index(drop=True)

# Feature: numeric time index
monthly_sales['t'] = np.arange(len(monthly_sales))

# Feature: month of year (seasonality signal)
monthly_sales['month'] = monthly_sales['YearMonth'].dt.month

print(monthly_sales.head(10))
print(f"\nTotal months of data: {len(monthly_sales)}")
```

```python
# Use last 6 months as test set
split = len(monthly_sales) - 6
train = monthly_sales[:split]
test  = monthly_sales[split:]

features = ['t', 'month']

X_train = train[features]
y_train = train['Sales']
X_test  = test[features]
y_test  = test['Sales']

# Linear Regression model
model = LinearRegression()
model.fit(X_train, y_train)

# Predictions
test = test.copy()
test['Predicted'] = model.predict(X_test)

# Also predict on train for full chart
train = train.copy()
train['Predicted'] = model.predict(X_train)
```

```python
mae  = mean_absolute_error(y_test, test['Predicted'])
rmse = np.sqrt(mean_squared_error(y_test, test['Predicted']))
mape = np.mean(np.abs((y_test - test['Predicted']) / y_test)) * 100

print("=" * 40)
print("  MODEL EVALUATION — TEST SET")
print("=" * 40)
print(f"  MAE  : ${mae:,.2f}")
print(f"  RMSE : ${rmse:,.2f}")
print(f"  MAPE : {mape:.2f}%")
print("=" * 40)
```

```python
fig, axes = plt.subplots(3, 1, figsize=(12, 14))
fig.suptitle('Sales Forecasting Dashboard — Superstore', 
             fontsize=16, fontweight='bold', y=0.98)

# --- Chart 1: Actual vs Predicted ---
ax1 = axes[0]
ax1.plot(train['YearMonth'], train['Sales'], 
         color='steelblue', label='Actual (train)', linewidth=2)
ax1.plot(test['YearMonth'], y_test, 
         color='steelblue', linestyle='--', label='Actual (test)', linewidth=2)
ax1.plot(test['YearMonth'], test['Predicted'], 
         color='tomato', label='Predicted', linewidth=2, marker='o', markersize=5)
ax1.axvline(test['YearMonth'].iloc[0], color='gray', linestyle=':', linewidth=1.5, label='Train/Test split')
ax1.set_title('Actual vs Predicted Monthly Sales', fontsize=13)
ax1.set_ylabel('Sales ($)')
ax1.legend()
ax1.xaxis.set_major_formatter(mdates.DateFormatter('%b %Y'))
ax1.tick_params(axis='x', rotation=30)
ax1.grid(True, alpha=0.3)

# --- Chart 2: Monthly Seasonality ---
ax2 = axes[1]
seasonal = df.groupby(df['Order Date'].dt.month)['Sales'].mean()
month_names = ['Jan','Feb','Mar','Apr','May','Jun',
               'Jul','Aug','Sep','Oct','Nov','Dec']
bars = ax2.bar(month_names, seasonal.values, color='steelblue', edgecolor='white')
ax2.set_title('Average Sales by Month (Seasonality)', fontsize=13)
ax2.set_ylabel('Avg Sales ($)')
ax2.grid(True, axis='y', alpha=0.3)
# Highlight the top month
top_idx = seasonal.values.argmax()
bars[top_idx].set_color('tomato')
ax2.annotate(f'Peak: {month_names[top_idx]}', 
             xy=(top_idx, seasonal.values[top_idx]),
             xytext=(top_idx + 0.5, seasonal.values[top_idx] * 1.02),
             fontsize=9, color='tomato')

# --- Chart 3: Error Analysis ---
ax3 = axes[2]
errors = y_test.values - test['Predicted'].values
ax3.bar(test['YearMonth'], errors, color=['tomato' if e < 0 else 'seagreen' for e in errors])
ax3.axhline(0, color='black', linewidth=1)
ax3.set_title('Forecast Error by Month (Actual − Predicted)', fontsize=13)
ax3.set_ylabel('Error ($)')
ax3.xaxis.set_major_formatter(mdates.DateFormatter('%b %Y'))
ax3.tick_params(axis='x', rotation=30)
ax3.grid(True, axis='y', alpha=0.3)

plt.tight_layout()
plt.savefig('sales_forecast_dashboard.png', dpi=150, bbox_inches='tight')
plt.show()
print("Chart saved as sales_forecast_dashboard.png")
```

```python
last_t = monthly_sales['t'].max()
last_date = monthly_sales['YearMonth'].max()

future_dates = pd.date_range(last_date + pd.offsets.MonthBegin(1), periods=6, freq='MS')
future_df = pd.DataFrame({
    'YearMonth': future_dates,
    't': np.arange(last_t + 1, last_t + 7),
    'month': future_dates.month
})
future_df['Predicted'] = model.predict(future_df[features])

# Plot future forecast
plt.figure(figsize=(12, 5))
plt.plot(monthly_sales['YearMonth'], monthly_sales['Sales'], 
         color='steelblue', label='Historical Sales', linewidth=2)
plt.plot(future_df['YearMonth'], future_df['Predicted'], 
         color='tomato', label='6-Month Forecast', 
         linewidth=2, linestyle='--', marker='o', markersize=6)
plt.axvline(future_dates[0], color='gray', linestyle=':', linewidth=1.5)
plt.title('6-Month Sales Forecast', fontsize=14, fontweight='bold')
plt.ylabel('Sales ($)')
plt.legend()
plt.xticks(rotation=30)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('future_forecast.png', dpi=150, bbox_inches='tight')
plt.show()

print("\nFuture Forecast:")
print(future_df[['YearMonth','Predicted']].to_string(index=False))
```
